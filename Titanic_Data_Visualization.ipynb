# Data Visualization with Matplotlib & Seaborn

**InternCircle Data Science Task 2**

This notebook visualizes the Titanic dataset using Matplotlib and Seaborn to identify patterns in passenger survival, demographics, fares, and passenger class.


## 1. Import Libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
```

## 2. Load and Prepare the Dataset


```python
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

# Handle missing values needed for visualization
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

print('Shape:', df.shape)
df.head()
```

## 3. Survival Count Bar Chart


```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='Survived')
plt.title('Titanic Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')
plt.show()
```

## 4. Survival by Gender


```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='Sex', hue='Survived')
plt.title('Survival Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

## 5. Survival Rate by Passenger Class


```python
class_rate = df.groupby('Pclass')['Survived'].mean().reset_index()

plt.figure(figsize=(7, 5))
sns.barplot(data=class_rate, x='Pclass', y='Survived')
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.ylim(0, 1)
plt.show()
```

## 6. Age Distribution Histogram


```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Age', bins=30, kde=True)
plt.title('Age Distribution of Titanic Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.show()
```

## 7. Fare Distribution by Passenger Class


```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Pclass', y='Fare')
plt.title('Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

## 8. Age Distribution by Survival


```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Survived', y='Age')
plt.title('Age Distribution by Survival Status')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Age')
plt.show()
```

## 9. Age vs Fare Scatter Plot


```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived', alpha=0.7)
plt.title('Age vs Fare by Survival Status')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

## 10. Correlation Heatmap


```python
numeric_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
corr = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
```

## 11. Key Visual Insights

- The survival count shows that more passengers did not survive than survived.
- Survival patterns differ substantially by gender.
- Passenger class is associated with different survival rates.
- The age histogram shows the distribution of passenger ages and highlights the broad range of ages on board.
- Fare values vary considerably across passenger classes, with higher variability and generally higher fares in premium classes.
- The scatter plot helps compare age, fare, and survival status.
- The correlation heatmap summarizes relationships among the main numerical variables.


## Conclusion

Matplotlib and Seaborn were used to transform the Titanic dataset into informative visualizations, including bar charts, histograms, box plots, scatter plots, and a correlation heatmap. These visualizations make passenger demographics, fares, class, and survival patterns easier to understand.
